```python
# 1. Import Libraries
import pandas as pd
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

DATASET_PATH = 'data/enron/enron_spam_data.csv'

df = pd.read_csv(DATASET_PATH)

df = df.dropna(subset=['Message'])
df['Subject'] = df['Subject'].fillna('')

# TODO Move the preprocessing methods to module TextProcessor
def clean_text(text):
        """Cleans raw text by removing punctuation, links, numbers, and stopwords."""
        if not isinstance(text, str):
            return ""
        text = text.lower()
        text = re.sub(r"http\S+|www\S+", "", text)  # remove URLs
        text = re.sub(r"[^\w\s]", "", text)  # remove punctuation
        text = re.sub(r"\d+", "", text)  # remove numbers
        tokens = text.split()
        tokens = [t for t in tokens if t not in ENGLISH_STOP_WORDS]
        return " ".join(tokens)


# Clean just the message
df['cleaned_text'] = df['Message'].apply(clean_text)

df['combined'] = (df['Subject'] + ' ' + df['Message']).apply(clean_text)

# Text length
df['text_length'] = df['Message'].apply(len)

# Word count
df['word_count'] = df['Message'].apply(lambda x: len(str(x).split()))

# Has link
df['has_link'] = df['Message'].apply(lambda x: int('http' in str(x).lower() or 'www' in str(x).lower()))

# Has dollar sign
df['has_dollar'] = df['Message'].apply(lambda x: int('$' in str(x)))

df.to_csv('data/enron/enron_preprocessed.csv', index=False)
print("Preprocessed data saved to: data/enron/enron_preprocessed.csv")
```
